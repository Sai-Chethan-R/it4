```python
"""
Task 4: Data Science Tool Mastery Project
Objective: Master industry-standard tools (Pandas, Seaborn, Scikit-Learn) 
and demonstrate proficiency through a practical machine learning portfolio piece.
"""

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

```python
# =============================================================================
# Tool 1: Pandas (Data Manipulation & Structuring)
# Generating a synthetic dataset to simulate a customer churn scenario
# =============================================================================
print("--- Initializing Pandas Data Structure ---")
X, y = make_classification(n_samples=1000, n_features=5, n_informative=3, 
                           random_state=42, weights=[0.7, 0.3])

# Creating a DataFrame to showcase tabular data mastery
df = pd.DataFrame(X, columns=['Usage_Time', 'Interaction_Score', 'Support_Tickets', 
                              'Account_Age', 'Recent_Activity'])
df['Churn_Status'] = y

print(df.head())
```

```python
# =============================================================================
# Tool 2: Seaborn & Matplotlib (Data Visualization Mastery)
# Creating a pairplot to visually analyze feature relationships
# =============================================================================
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 6))

# Plotting the distribution of Support Tickets vs Usage Time, segmented by Churn
sns.scatterplot(data=df, x='Usage_Time', y='Support_Tickets', 
                hue='Churn_Status', palette='coolwarm', alpha=0.7)
plt.title('Tool Mastery: Seaborn Visualization (Usage vs Support Tickets)')
plt.show()
```

```python
# =============================================================================
# Tool 3: Scikit-Learn (Machine Learning & Predictive Modeling)
# Building, training, and evaluating a Random Forest Classifier
# =============================================================================
print("\n--- Initializing Scikit-Learn Pipeline ---")

# Splitting data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(df.drop('Churn_Status', axis=1), 
                                                    df['Churn_Status'], 
                                                    test_size=0.2, 
                                                    random_state=42)

# Instantiating and training the model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Generating predictions
predictions = rf_model.predict(X_test)

# Evaluating tool proficiency via statistical metrics
print("\n--- Model Evaluation (Confusion Matrix) ---")
print(confusion_matrix(y_test, predictions))

print("\n--- Model Evaluation (Classification Report) ---")
print(classification_report(y_test, predictions))
```
